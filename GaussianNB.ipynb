```python
# imports

import pandas as pd

#Gaussian naive bayes
from sklearn.naive_bayes import GaussianNB

#80%- training and 20% - testing

from sklearn.model_selection import train_test_split

# model accuracy for categories finding: use accuracy score

from sklearn.metrics  import accuracy_score

# to convert output labels into numbers(0 or 1)-label encoder
from sklearn.preprocessing import LabelEncoder

```

```python
df=pd.read_csv("/content/diabetes_prediction_dataset.csv")
df
```

```python
# input and output

#input(x)

x=df.select_dtypes(include=['int64','float64']).drop(columns=['diabetes'])

#output(y)
y=df['diabetes']

# df.columns
```

```python
model=GaussianNB()

# to assign output labels such as 0 for no diabetes and 1 for diabetes
le=LabelEncoder()

y_val=le.fit_transform(y)

```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y_val)
# now splitting x,y_val - input and output data for testing and trianing

# model training
model.fit(x_train,y_train)


```

```python
# testing
y_pred=model.predict(x_test)
# y_pred
```

```python
# accuracy score --> it compare actual and predicted values

acc=accuracy_score(y_test,y_pred)

print("Accuracy score: ",acc)

#accuracy:0.9 -> excellent model
```

```python
# user prediction inputs
# input features(x): 'age', 'hypertension', 'heart_disease','bmi',
# 'HbA1c_level', 'blood_glucose_level

age=int(input("Enter age: "))
hypertension=int(input("Enter hypertension: "))
heart_disease=int(input("Enter heart disease: "))
bmi=float(input("Enter bmi: "))
HB=float(input("Enter HbA1c_level: "))
blood_glucose=int(input("Enter blood glucose level: "))


user=pd.DataFrame({'age':[age],'hypertension':[hypertension],
                   'heart_disease':[heart_disease],'bmi':[bmi],
                   'HbA1c_level':[HB],
                   'blood_glucose_level':[blood_glucose]})
# user
# predict output
prediction=model.predict(user)

if prediction[0]==1:
  print("You have Diabetes")
else:
  print("You have no diabetes")


# output(y): 'diabetes' --> model predict
# df.columns
```
