```python
# Question 1:
# Why would you choose Multiple Linear Regression
# instead of Simple Linear Regression?

from sklearn.linear_model import LinearRegression

# Multiple input features
# Area, Bedrooms, Bathrooms
X = [
    [1000, 2, 1],
    [1500, 3, 2],
    [2000, 3, 2],
    [2500, 4, 3],
    [3000, 4, 3]
]

# House prices
y = [2000000, 3000000, 4000000, 5000000, 6000000]

# Create Multiple Linear Regression model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict price
prediction = model.predict([[1800, 3, 2]])

print("Predicted House Price:", prediction[0])
```

```python
# Question 2:
# Can adding more features always improve the model?

from sklearn.linear_model import LinearRegression

# Meaningful features
X = [
    [1000, 2],
    [1500, 3],
    [2000, 3],
    [2500, 4],
    [3000, 4]
]

y = [2000000, 3000000, 4000000, 5000000, 6000000]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[1800, 3]])

print("Predicted House Price:", prediction[0])
```

```python
# Question 3:
# How do you identify the most important feature
# in a Multiple Linear Regression model?

from sklearn.linear_model import LinearRegression

X = [
    [1000, 2, 1],
    [1500, 3, 2],
    [2000, 3, 2],
    [2500, 4, 3],
    [3000, 4, 3]
]

y = [2000000, 3000000, 4000000, 5000000, 6000000]

model = LinearRegression()

model.fit(X, y)

# Display coefficients
print("Area coefficient:", model.coef_[0])
print("Bedrooms coefficient:", model.coef_[1])
print("Bathrooms coefficient:", model.coef_[2])
```

```python
# Question 4:
# What is Multicollinearity?

import pandas as pd

data = {
    "Area": [1000, 1500, 2000, 2500, 3000],
    "Bedrooms": [2, 3, 3, 4, 4],
    "Bathrooms": [1, 2, 2, 3, 3]
}

df = pd.DataFrame(data)

# Find correlation between features
correlation = df.corr()

print(correlation)
```

```python
# Question 5:
# Can Multiple Linear Regression work with only one feature?

from sklearn.linear_model import LinearRegression

# Only one feature
X = [[1], [2], [3], [4], [5]]

y = [45, 50, 55, 60, 65]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted Marks:", prediction[0])
```
