```python
import pandas as pd
# multinomial naive bayes import
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/StudentsPerformance (1).csv")
df
```

```python
df["average score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
) / 3

df
```

```python
df["result"] = df["average score"].apply(
    lambda x: 1 if x >= 50 else 0
)

df[["average score", "result"]].head()
```

```python
# input and output

x = (
    df["gender"] + " " +
    df["race/ethnicity"] + " " +
    df["parental level of education"] + " " +
    df["lunch"] + " " +
    df["test preparation course"]
)

y = df["result"]
```

```python
vector = CountVectorizer()

x_val = vector.fit_transform(x)
```

```python
model = MultinomialNB()
```

```python
model.fit(x_val, y)
```

```python
print("Model trained successfully")
```

```python
while True:

    gender = input("Enter gender: ")
    race = input("Enter race/ethnicity: ")
    education = input("Enter parental education: ")
    lunch = input("Enter lunch type: ")
    preparation = input("Enter test preparation course: ")

    student_data = (
        gender + " " +
        race + " " +
        education + " " +
        lunch + " " +
        preparation
    )

    test_student = vector.transform([student_data])

    prediction = model.predict(test_student)

    print("Predicted Result:", prediction[0])

    if prediction[0] == 1:
        print("Student is PASS")
    else:
        print("Student is FAIL")
```
