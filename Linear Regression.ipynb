```python
# Question 1:
# Suppose your company wants to predict the monthly electricity bill
# of a customer. Which Machine Learning algorithm would you choose? Why?

from sklearn.linear_model import LinearRegression

# Electricity usage
X = [[100], [200], [300], [400], [500]]

# Electricity bill
y = [500, 900, 1300, 1700, 2100]

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict electricity bill for 350 units
prediction = model.predict([[350]])

print("Predicted Electricity Bill:", prediction[0])
```

```python
# Question 2:
# Why can't we use Logistic Regression to predict house prices?

from sklearn.linear_model import LinearRegression

# House area
X = [[1000], [1500], [2000], [2500]]

# House prices
y = [2000000, 3000000, 4000000, 5000000]

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict house price for 1800 sq.ft
prediction = model.predict([[1800]])

print("Predicted House Price:", prediction[0])
```

```python
# Question 3:
# What does the slope represent in Linear Regression?

from sklearn.linear_model import LinearRegression

# Study hours
X = [[1], [2], [3], [4], [5]]

# Marks
y = [45, 50, 55, 60, 65]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Get slope
slope = model.coef_[0]

print("Slope:", slope)
```

```python
# Question 4:
# What is the purpose of the R² Score?

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Study hours
X = [[1], [2], [3], [4], [5]]

# Marks
y = [45, 50, 55, 60, 65]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict values
y_pred = model.predict(X)

# Calculate R² Score
r2 = r2_score(y, y_pred)

print("R² Score:", r2)
```

```python
# Question 5:
# If the R² Score is 0.25, what does it indicate?

r2 = 0.25

# Convert R² into percentage
percentage = r2 * 100

print("R² Score:", r2)
print("Variation explained:", percentage, "%")
print("Variation not explained:", 100 - percentage, "%")
```

```python
# Question 6:
# How would you improve a poor Linear Regression model?

from sklearn.linear_model import LinearRegression

# Multiple features:
# Study Hours and Attendance
X = [
    [1, 60],
    [2, 65],
    [3, 70],
    [4, 80],
    [5, 90]
]

# Marks
y = [45, 50, 55, 65, 75]

# Create Multiple Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict marks for 6 hours study and 95% attendance
prediction = model.predict([[6, 95]])

print("Predicted Marks:", prediction[0])
```
